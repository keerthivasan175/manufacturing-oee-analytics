```python
# 02 - Load cleaned data into MySQL
import pandas as pd
from sqlalchemy import create_engine, text

PASSWORD = "YOUR_PASSWORD"   # replace locally, never commit your real password
engine = create_engine(f"mysql+pymysql://root:{PASSWORD}@localhost:3307/manufacturing_oee_db")

clean = pd.read_csv("../data/cleaned/fact_production_runs_clean.csv", keep_default_na=False)
with engine.begin() as conn:
    conn.execute(text("TRUNCATE TABLE fact_production_runs"))
clean.to_sql("fact_production_runs", engine, if_exists="append", index=False, chunksize=5000, method="multi")
```

```python
print(pd.read_sql("SELECT COUNT(*) AS rows_loaded FROM fact_production_runs", engine))
```
